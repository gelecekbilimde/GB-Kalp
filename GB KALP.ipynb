```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Setup the figure
fig, ax = plt.subplots(figsize=(5.5, 9), facecolor='black')
ax.set_facecolor('black')

ax.set_aspect('equal') 

# Center the axes 
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

# Color the axes and numbers white
ax.spines['left'].set_color('white')
ax.spines['bottom'].set_color('white')
ax.tick_params(axis='x', colors='white')
ax.tick_params(axis='y', colors='white')

# Set grid limits
ax.set_xlim(-2.5, 2.5)
ax.set_ylim(-1.5, 2.5)
ax.set_xticks([-2, -1, 1, 2])
ax.set_yticks([-1, 1, 2])

fig.text(0.5, 0.80, 'Sevgililer Gününüz Kutlu Olsun!', color='#ff4d4d', 
         fontsize=22, ha='center', va='center')

fig.text(0.5, 0.76, 'BİLİMİ SEVİYORUZ! ❤️', color='#00b73b', 
         fontsize=18, ha='center', va='center', fontweight='bold')

# The equation
eq_text = r'$y = x^{\frac{2}{3}} + 0.9 \sin(kx) \sqrt{3 - x^2}$'
fig.text(0.5, 0.22, eq_text, color='#096882', 
         fontsize=18, ha='center', va='center', fontweight='bold')

# The dynamic K-value
k_text = fig.text(0.5, 0.15, '', color='#ff4d4d', 
                  fontsize=18, ha='center', va='center', fontfamily='serif')

# Math
x = np.linspace(-np.sqrt(3), np.sqrt(3), 5000)

line, = ax.plot([], [], color='#ff3333', lw=1.5)

def init():
    line.set_data([], [])
    k_text.set_text('')
    return line, k_text

def update(frame):
    k = frame
    y = np.cbrt(x**2) + 0.9 * np.sin(k * x) * np.sqrt(3 - x**2)
    
    line.set_data(x, y)
    k_text.set_text(f'k = {k:.2f}')
    
    return line, k_text

# Animation
k_values = np.linspace(0, 100, 600) # duration

ani = FuncAnimation(fig, update, frames=k_values, init_func=init, blit=True)

ani.save('GB Kalp.mp4', writer='ffmpeg', fps=30, dpi=200)
```
